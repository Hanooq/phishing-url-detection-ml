## 1. Import libraries and load the dataset

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

data = pd.read_csv('simulated_phishing_url_dataset.csv')
print(data.shape)
data.head()
```

## 2. Understand the data

`label = 1` means simulated phishing; `label = 0` means simulated legitimate.

```python
print(data.info())
print(data['label'].value_counts())
print(data.isnull().sum())
```

## 3. Split features and target

```python
X = data.drop(columns=['label'])
y = data['label']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
print('Training rows:', len(X_train), '| Testing rows:', len(X_test))
```

## 4. Train a Random Forest model

```python
model = RandomForestClassifier(n_estimators=120, max_depth=8, random_state=42, class_weight='balanced')
model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

## 5. Evaluate the model

These scores describe only the simulated dataset.

```python
print('Accuracy:', round(accuracy_score(y_test, predictions), 3))
print(classification_report(y_test, predictions, target_names=['Legitimate', 'Phishing']))
print('Confusion matrix:\n', confusion_matrix(y_test, predictions))
```

## 6. View feature importance

```python
importance = pd.Series(model.feature_importances_, index=X.columns).sort_values(ascending=False)
importance
```

## 7. Try one example

The values below are illustrative feature values, not a real URL scan.

```python
example = pd.DataFrame([{'url_length':72,'num_dots':4,'num_hyphens':3,'has_at_symbol':1,'uses_ip_address':0,'uses_https':0,'num_digits':8,'suspicious_word_count':2,'uses_url_shortener':0}])
result = model.predict(example)[0]
print('Potentially phishing' if result == 1 else 'Likely legitimate')
```

## Conclusion

This notebook demonstrates a basic supervised-learning workflow. A real deployment requires a verified public dataset, careful validation, and security testing. A model prediction must never be treated as a guarantee that a site is safe.
